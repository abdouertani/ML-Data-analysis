### Data Preparation for LLM Fine-tuning

Now that we have a larger synthetic dataset (`df_cybersecurity_large`), we need to prepare it for fine-tuning a Transformer model. This involves several key steps:

1.  **Label Encoding:** Convert our 'legitimate' and 'phishing' labels into numerical IDs.
2.  **Dataset Split:** Divide the dataset into training and testing sets to evaluate the model's performance on unseen data.
3.  **Tokenization:** Convert the text messages into a format (numerical tokens and attention masks) that the Transformer model can process. We'll use a tokenizer corresponding to our chosen pre-trained model.

```python
from sklearn.model_selection import train_test_split
from transformers import AutoTokenizer
import torch
from datasets import Dataset

# 1. Label Encoding
# Map labels to integers
label_mapping = {'legitimate': 0, 'phishing': 1}
df_cybersecurity_large['labels'] = df_cybersecurity_large['label'].map(label_mapping)

# 2. Dataset Split
# Split the DataFrame into training and testing sets
train_df, test_df = train_test_split(df_cybersecurity_large, test_size=0.2, stratify=df_cybersecurity_large['labels'], random_state=42)

# Convert pandas DataFrames to Hugging Face Dataset objects
train_dataset = Dataset.from_pandas(train_df[['text', 'labels']])
test_dataset = Dataset.from_pandas(test_df[['text', 'labels']])

# 3. Tokenization
# Choose a pre-trained model and load its tokenizer
# DistilBERT is a good choice for efficiency and performance
model_checkpoint = 'distilbert-base-uncased'
tokenizer = AutoTokenizer.from_pretrained(model_checkpoint)

def tokenize_function(examples):
    return tokenizer(examples['text'], truncation=True, padding='max_length', max_length=128)

# Apply tokenization to the datasets
tokenized_train_dataset = train_dataset.map(tokenize_function, batched=True)
tokenized_test_dataset = test_dataset.map(tokenize_function, batched=True)

# Remove the original 'text' and '__index_level_0__' columns as they are no longer needed after tokenization
tokenized_train_dataset = tokenized_train_dataset.remove_columns(['text', '__index_level_0__'])
tokenized_test_dataset = tokenized_test_dataset.remove_columns(['text', '__index_level_0__'])

# Set the format to PyTorch tensors
tokenized_train_dataset.set_format('torch')
tokenized_test_dataset.set_format('torch')

print("\n--- Data Preparation Summary ---")
print(f"Training dataset size: {len(tokenized_train_dataset)}")
print(f"Test dataset size: {len(tokenized_test_dataset)}")
print("First tokenized training example:")
print(tokenized_train_dataset[0])
print("Label mapping:", label_mapping)
```

### Loading the Model and Setting up Training

Now we will load our chosen pre-trained Transformer model (`DistilBERT` for this task). We'll also define the evaluation metrics (accuracy, F1-score) that will be used to monitor the model's performance during training and on the test set. Finally, we'll configure the training parameters using `TrainingArguments`.

```python
from transformers import AutoModelForSequenceClassification, TrainingArguments, Trainer
import evaluate
import numpy as np

# 1. Load Pre-trained Model
# We'll use DistilBERT for sequence classification with 2 labels (legitimate, phishing)
model = AutoModelForSequenceClassification.from_pretrained(model_checkpoint, num_labels=len(label_mapping))

# 2. Define Evaluation Metrics
# We'll use accuracy and F1-score
metric = evaluate.load('f1') # F1-score is good for binary classification, especially with potential imbalance
accuracy_metric = evaluate.load('accuracy')

def compute_metrics(eval_pred):
    predictions, labels = eval_pred
    predictions = np.argmax(predictions, axis=1)
    f1 = metric.compute(predictions=predictions, references=labels, average='binary')['f1']
    accuracy = accuracy_metric.compute(predictions=predictions, references=labels)['accuracy']
    return {'f1': f1, 'accuracy': accuracy}

# 3. Configure Training Arguments
# Define where to save the model, batch size, number of epochs, etc.
training_args = TrainingArguments(
    output_dir='./results',
    eval_strategy='epoch',
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    num_train_epochs=3,
    weight_decay=0.01,
    save_total_limit=2, # Only save the last 2 checkpoints
    metric_for_best_model='f1', # Use f1 score to determine the best model
    push_to_hub=False, # We are not pushing the model to Hugging Face Hub
)

# 4. Initialize the Trainer
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_train_dataset,
    eval_dataset=tokenized_test_dataset,
    compute_metrics=compute_metrics,
)

print("\n--- Model and Trainer Setup Complete ---")
print(f"Model loaded: {model_checkpoint}")
print(f"Number of labels: {model.config.num_labels}")
print(f"Training arguments configured for {training_args.num_train_epochs} epochs.")
print("Trainer initialized. Ready to start training!")
```

### Training the Model

We will now fine-tune the `DistilBERT` model on our tokenized training dataset. The training process will iterate over the data for the specified number of epochs, optimizing the model's parameters to improve its classification accuracy.

```python
# Start training
trainer.train()

print("\n--- Training Complete ---")
```

### Evaluating the Model

After training, it's crucial to evaluate the model's performance on the unseen test dataset to get an unbiased estimate of its generalization capabilities. We'll use the accuracy and F1-score metrics we defined earlier.

```python
# Evaluate the model on the test set
results = trainer.evaluate()

print("\n--- Evaluation Results ---")
for key, value in results.items():
    print(f"{key}: {value:.4f}")

# Optional: Display some predictions from the test set
print("\n--- Sample Predictions from Test Set ---")
predictions, labels, _ = trainer.predict(tokenized_test_dataset)
predictions = np.argmax(predictions, axis=1)

# Inverse mapping for labels
inverse_label_mapping = {v: k for k, v in label_mapping.items()}

for i in range(5):
    original_text = test_df.iloc[i]['text']
    true_label = inverse_label_mapping[labels[i].item()]
    predicted_label = inverse_label_mapping[predictions[i]]
    print(f"\nText: {original_text}")
    print(f"True Label: {true_label}")
    print(f"Predicted Label: {predicted_label}")
```
